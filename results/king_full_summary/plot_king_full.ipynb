```python
import pandas as pd
import plotly.express as px
```

```python
king_coef = pd.read_csv('king_full_kinship_results.csv')
king_coef
```

```python
king_seg = pd.read_csv('king_full_segment_results.csv')
king_seg
```

```python
king_seg['RC'].value_counts()
```

## First Order Relationships (Parent-Child and Sibling Relationships)

```python
first_order = ['Parent Child', 'Full sibling', 'Half Sibling']
king_seg_first_order = king_seg[king_seg['RC'].isin(first_order)]

fig = px.violin(king_seg_first_order, y="PropIBD", x="RC", color="pop", box=True, points="all", 
                hover_data=king_seg_first_order.columns, template='simple_white',
                title="First Order Kinship Coefficients using KING PropIBD")
fig.show()
```

```python
first_order = ['Parent Child', 'Full sibling', 'Half Sibling']
king_coef_first_order = king_coef[king_coef['RC'].isin(first_order)]

fig = px.violin(king_coef_first_order, y="Kinship", x="RC", color="pop", box=True, points="all", 
                hover_data=king_coef_first_order.columns, template='simple_white',
                title="First Order Kinship Coefficients using KING Kinship Coefficients")
fig.show()
```
