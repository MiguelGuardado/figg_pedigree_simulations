```python
import pandas as pd
import plotly.express as px
```

```python
ibis_coef = pd.read_csv('ibis_full_kinship_results.csv')
ibis_coef
```

```python
ibis_seg = pd.read_csv('ibis_full_segment_results.csv')
ibis_seg
```

```python
ibis_seg_total_ibd = (
    ibis_seg.groupby(["Individual1", "Individual2"], as_index=False)
      .agg(
          Fam_ID=("Fam_ID", "first"),
          relationship=("RC", "first"),        # "Parent Child", etc.
          pop=("pop", "first"),
          total_ibd_cM=("genetic_seg_length", "sum"),
          total_markers=("marker_count", "sum"),
          n_segments=("genetic_seg_length", "size"),
      )
)
ibis_seg_total_ibd
```

## First Order Relationships (Parent-Child and Sibling Relationships)

```python
first_order = ['Parent Child', 'Full sibling', 'Half Sibling']
ibis_coef_first_order = ibis_coef[ibis_coef['RC'].isin(first_order)]

fig = px.violin(ibis_coef_first_order, y="Kinship_Coefficient", x="RC", color="pop", box=True, points="all", 
                hover_data=ibis_coef_first_order.columns, template='simple_white',
                title="First Order Kinship Coefficients using IBIS Kinship Coefficients")
fig.show()
```

# Full Cousin Relationships

```python
full_cousin = ['Full First Cousin', 'Full First Cousin', 'Full First Cousin', 'Full Second Cousin', 'Full Third Cousin', 'Full Fourth Cousin', 'Full Fifth Cousin']
ibis_coef_full_cousin = ibis_coef[ibis_coef['RC'].isin(full_cousin)]

fig = px.violin(ibis_coef_full_cousin, y="Kinship_Coefficient", x="RC", color="pop", box=True, points="all", 
                hover_data=ibis_coef_full_cousin.columns, template='simple_white',
                title="Full Cousin Kinship Coefficients using IBIS Kinship Coefficients")
fig.show()
```
