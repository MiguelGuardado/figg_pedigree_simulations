```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
hap_ibd_seg = pd.read_csv(f'hap_ibd_snp_panel_results.csv')
hap_ibd_seg
```

```python
keys = ["pop", "seed", "Fam_ID", "ID1", "ID2"]

# Order-invariant pair; swap the haplotype labels along with the IDs
swap = (hap_ibd_seg["ID1"] > hap_ibd_seg["ID2"]).to_numpy()
hap_ibd_seg.loc[swap, ["ID1", "ID2", "ID1_H", "ID2_H"]] = hap_ibd_seg.loc[swap, ["ID2", "ID1", "ID2_H", "ID1_H"]].to_numpy()

hap_ibd_seg_new = (
    hap_ibd_seg.groupby(keys, sort=False, observed=True)
       .agg(
           RT=("RT", "first"),
           RC=("RC", "first"),
           MC=("MC", "first"),
           GD=("GD", "first"),
           n_seg=("CM_Length", "size"),
           total_hap_cM=("CM_Length", "sum"),
           max_seg_cM=("CM_Length", "max"),
       )
       .reset_index()
)
hap_ibd_seg_new
```

## First Order

```python
fo_rels = ['Parent Child', 'Full sibling', 'Half Sibling']

hap_ibd_fo = hap_ibd_seg_new[hap_ibd_seg_new['RC'].isin(fo_rels)]
hap_ibd_fo

fig = px.violin(hap_ibd_fo, y="total_hap_cM", x="RC", color='pop',
                hover_data=hap_ibd_fo.columns, template='simple_white')
fig.update_layout(xaxis_title=' ')
fig.show()
```
