```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
king_kinship = pd.read_csv('king_full_kinship_results.csv')
king_kinship
```

# KING Kinship

## First Order Relationships

```python
fo_rels = ['Parent Child', 'Full sibling', 'Half Sibling']
king_kinship_fo = king_kinship[king_kinship['RC'].isin(fo_rels)]

expected_fo = {
    'Parent Child': 1/4,   # 0.25
    'Full sibling': 1/4,   # 0.25
    'Half Sibling': 1/8,   # 0.125
}

fig = px.violin(king_kinship_fo, y="Kinship", x="RC", color='pop',
                hover_data=king_kinship_fo.columns, template='simple_white',
                category_orders={'RC': fo_rels})

for i, rel in enumerate(fo_rels):
    y = expected_fo[rel]
    fig.add_shape(type='line', xref='x', x0=i-0.5, x1=i+0.5, yref='y', y0=y, y1=y,
                  line=dict(color='red', dash='dash', width=2))

fig.update_layout(xaxis_title=' ')
fig.update_layout(title='KING Kinship Coef, First Order, Full Genome')
fig.write_image('images/king_kinship_first_order_full_genome.png', scale=2)
fig.show()
```

### Cousin Relationships 

```python
cuz_rels = ['Full First Cousin', 'Full Second Cousin', 'Full Third Cousin',
            'Half Forth Cousin', 'Full Fifth Cousin']
king_kinship_cuz = king_kinship[king_kinship['RC'].isin(cuz_rels)]

expected_cuz = {
    'Full First Cousin': 1/16,
    'Full Second Cousin': 1/64,
    'Full Third Cousin': 1/256,
    'Half Forth Cousin': 1/2048,
    'Full Fifth Cousin': 1/4096,
}

fig = px.violin(king_kinship_cuz, y="Kinship", x="RC", color='pop',
                hover_data=king_kinship_cuz.columns, template='simple_white',
                category_orders={'RC': cuz_rels})

for i, rel in enumerate(cuz_rels):
    y = expected_cuz[rel]
    fig.add_shape(type='line', xref='x', x0=i-0.5, x1=i+0.5, yref='y', y0=y, y1=y,
                  line=dict(color='red', dash='dash', width=2))

fig.update_layout(xaxis_title=' ')
fig.update_layout(title='KING Kinship Coef, Cousins, Full Genome')
fig.write_image('images/king_kinship_cousin_full_genome.png', scale=2)
fig.show()
```

### Grandparent

```python
anc_rels = ['Grandparent', 'Great Grandparent', 'Great Great Grandparent',
            'Great Great Great Grandparent']

# your list had trailing spaces on two labels; strip them in the data so they match
king_kinship['RC'] = king_kinship['RC'].str.strip()

king_kinship_anc = king_kinship[king_kinship['RC'].isin(anc_rels)]

fig = px.violin(king_kinship_anc, y="Kinship", x="RC", color='pop',
                hover_data=king_kinship_anc.columns, template='simple_white',
                category_orders={'RC': anc_rels})

expected_anc = {
    'Grandparent':                    1/8,    # 0.125
    'Great Grandparent':              1/16,   # 0.0625
    'Great Great Grandparent':        1/32,   # 0.03125
    'Great Great Great Grandparent':  1/64,   # 0.015625
}

# one dashed line per relationship, spanning only its own category
for i, rel in enumerate(anc_rels):
    y = expected_anc[rel]
    fig.add_shape(type='line', xref='x', x0=i-0.5, x1=i+0.5, yref='y', y0=y, y1=y,
                  line=dict(color='red', dash='dash', width=2))

fig.update_layout(xaxis_title=' ')
fig.update_layout(title='KING Kinship Coef, Direct Ancestors, Full Genome')
fig.write_image('images/king_kinship_direct_ancestors_full_genome.png', scale=2)
fig.show()
```
