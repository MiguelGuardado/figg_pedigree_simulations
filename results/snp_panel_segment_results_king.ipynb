```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
king_segment = pd.read_csv('king_snp_panel_segment_results.csv')

# strip stray whitespace from relationship labels so they match the lists below
king_segment['RC'] = king_segment['RC'].str.strip()

king_segment
```

```python
# KING only reports segment results for pairs it detects, so distant relationships
# have fewer pairs here than in the kinship results
king_segment['RC'].value_counts()
```

# KING Segment (PropIBD)

PropIBD = 2 × kinship coefficient, so every expected line is double the kinship version (e.g. Parent Child kinship 0.25 → PropIBD 0.50).

## First Order Relationships

```python
fo_rels = ['Parent Child', 'Full sibling', 'Half Sibling']

expected_fo = {
    'Parent Child': 1/2,   # 0.5   (kinship 0.25)
    'Full sibling': 1/2,   # 0.5   (kinship 0.25)
    'Half Sibling': 1/4,   # 0.25  (kinship 0.125)
}

king_segment_fo = king_segment[king_segment['RC'].isin(fo_rels)]

fig = px.violin(king_segment_fo, y="PropIBD", x="RC", color='pop',
                hover_data=king_segment_fo.columns, template='simple_white',
                category_orders={'RC': fo_rels})

# one dashed line per relationship, spanning only its own category
for i, rel in enumerate(fo_rels):
    y = expected_fo[rel]
    fig.add_shape(type='line', xref='x', x0=i-0.5, x1=i+0.5, yref='y', y0=y, y1=y,
                  line=dict(color='red', dash='dash', width=2))

fig.update_layout(xaxis_title=' ', yaxis_title='PropIBD')
fig.update_layout(title='KING PropIBD, First Order, SNP Panel')
fig.write_image('images/king_segment_first_order_snp_panel.png', scale=2)
fig.show()
```

## Direct Ancestors

```python
anc_rels = ['Grandparent', 'Great Grandparent', 'Great Great Grandparent',
            'Great Great Great Grandparent']

expected_anc = {
    'Grandparent':                    1/4,    # 0.25     (kinship 1/8)
    'Great Grandparent':              1/8,    # 0.125    (kinship 1/16)
    'Great Great Grandparent':        1/16,   # 0.0625   (kinship 1/32)
    'Great Great Great Grandparent':  1/32,   # 0.03125  (kinship 1/64)
}

king_segment_anc = king_segment[king_segment['RC'].isin(anc_rels)]

fig = px.violin(king_segment_anc, y="PropIBD", x="RC", color='pop',
                hover_data=king_segment_anc.columns, template='simple_white',
                category_orders={'RC': anc_rels})

# one dashed line per relationship, spanning only its own category
for i, rel in enumerate(anc_rels):
    y = expected_anc[rel]
    fig.add_shape(type='line', xref='x', x0=i-0.5, x1=i+0.5, yref='y', y0=y, y1=y,
                  line=dict(color='red', dash='dash', width=2))

fig.update_layout(xaxis_title=' ', yaxis_title='PropIBD')
fig.update_layout(title='KING PropIBD, Direct Ancestors, SNP Panel')
fig.write_image('images/king_segment_direct_ancestors_snp_panel.png', scale=2)
fig.show()
```
